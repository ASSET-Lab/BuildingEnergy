Convert the TGW global warming deltas provided at monthly intervals for the CONUS into a three column (year, scenario, warming) table for joining onto EnergyPlus simulation results. 

```python
from pathlib import Path 
import pandas as pd 

conus_monthly_warming_deltas_folder = Path("/Volumes/seas-mtcraig/data_sharing/Energy Burdens Under Climate Change/TGW Warming Deltas/CONUS_Monthly_Deltas")

files = {
    "rcp45cooler": "CONUS_Monthly_Delta_rcp45cooler.csv",
    "rcp45hotter": "CONUS_Monthly_Delta_rcp45hotter.csv",
    "rcp85cooler": "CONUS_Monthly_Delta_rcp85cooler.csv",
    "rcp85hotter": "CONUS_Monthly_Delta_rcp85hotter.csv",
    }

conus_monthly_warming_deltas = []

for scenario, deltas_file in files.items():
    scenario_conus_monthly_deltas = pd.read_csv(Path(conus_monthly_warming_deltas_folder, deltas_file))

    scenario_conus_monthly_deltas = scenario_conus_monthly_deltas.set_index(scenario_conus_monthly_deltas['Year'])
    scenario_conus_monthly_deltas = scenario_conus_monthly_deltas.transpose()

    scenario_conus_monthly_deltas = scenario_conus_monthly_deltas.iloc[1:]  # drop first row (old header)
    
    # for every thing but the first column 
    for year in scenario_conus_monthly_deltas.columns.to_list():
        new_df = pd.DataFrame()

        new_df['Month'] = scenario_conus_monthly_deltas.index
        new_df['Weather Scenario'] = scenario
        new_df['Year'] = year
        new_df['CONUS Climate Change [Celsius]'] = scenario_conus_monthly_deltas[year].values

        conus_monthly_warming_deltas.append(new_df)

conus_monthly_warming_deltas = pd.concat(conus_monthly_warming_deltas, ignore_index=True)
conus_monthly_warming_deltas.to_csv(Path(conus_monthly_warming_deltas_folder, "CONUS_Monthly_Delta_all_scenarios.csv"), index=False)
```
